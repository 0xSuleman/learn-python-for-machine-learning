**The Dumb Agent vs Smart Agent**

```python
# A simple reflex agent have no memory; it only looks at what is right in front of its face and follows basic if/else rules
def dumbVacuum(location, dirty):
  if dirty:
    return "Clean"
  if location == "A":
    return "Move to B"
  return "Move to A"

# A model based agent have memory, so it remembers if he cleaned A previously and does not oscillate between A and B if both are cleaned
class SmartVacuum:
  def __init__(self):
    self.memory = {"A" : None, "B" : None}
  def action(self, location, dirty):

    if dirty:
      self.memory[location] = "Dirty"
    else:
      self.memory[location] = "Clean"

    if dirty:
      self.memory[location] = "Clean"
      return "Clean"

    if self.memory["A"] == "Clean" and self.memory["B"] == "Clean":
      return "Turn Off"

    if location == "A":
      return "Move to B"
    else:
      return "Move to A"
```

**Task 1: Traffic Light Agent**

```python
def trafficLightAgent(queues):
  nsCars = queues["NorthSouth"]
  ewCars = queues["EastWest"]

  if nsCars > ewCars:
    return "Green Signal at North South"
  elif ewCars > nsCars:
    return "Green Signal at East West"
  else:
    return "Keep the Current Signal"
```

**Task 3: Why a Reflex Agent Fails in Real Life**

A reflex agent looks only at one instant snapshot.  

Imagine an automatic car braking system that only measures distance to the car ahead:

Percept: distance = 15 meters.  

If the car is driving in a parking lot at 5 km/h, slamming the emergency brakes throws passengers forward unnecessarily.

If the car is on a highway at 120 km/h, failing to emergency brake causes a high-speed crash.

Because the simple reflex agent does not remember its own speed over time **(state)**, it cannot make the right decision from distance alone.

**Task 4: Classifying Environments for Delivery Robot**

Understanding the world a delivery robot travels through:

**Partially Observable:** The robot cannot see through brick walls or around blind corners (it only sees what its camera points at).

**Stochastic (Unpredictable):** A pedestrian might suddenly jump in front of the robot, or wheels might slip on wet mud (not 100% predictable).  

**Dynamic (Constantly Moving):** While the robot stops to calculate a path, traffic and pedestrians keep moving around it.  

**Continuous:** Distances, angles, and vehicle speeds are smooth real numbers, not grid boxes.

**Sequential:** Turning left down the wrong street now affects where the robot ends up 10 minutes later.
